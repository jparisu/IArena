# GoldMine

El juego propuesto para este curso es el juego de [GoldMine](https://iarena.readthedocs.io/en/latest/src/games/docs/goldmine.html)

Este juego consiste en un grid de casillas con coste, donde hay que ir excavando las distintas casillas para encontrar aquella que contiene oro.
Cada casilla tiene un coste de excavación, y el objetivo es encontrar el oro con el menor coste posible.
El juego requiere de encontrar caminos para moverse a lo largo del mapa, e idear una estrategia para elegir qué casillas excavar la siguiente.

```python
# CODE SOLUTION
AUTHORS = ["John Doe", "Jane Doe"]  # REPLACE WITH YOUR NAMES

# IMPORTANT: do not remove these imports
from iarena.gaming.goldmine import GoldMineMovement, GoldMinePosition, GoldMineRules, GoldMinePerfectPlayer
from iarena.playing.StudentPlayer import StudentPlayer

# Extra imports
import random

# Define player class
class MyPlayer(StudentPlayer):

    def starting_game(self, rules: GoldMineRules, player_index: int) -> None:
        self.best_player = GoldMinePerfectPlayer(seed=0)
        self.best_player.starting_game(rules, 0)

    def play(self, pos: GoldMinePosition) -> GoldMineMovement:
        return self.best_player.play(pos)



# Create player instance (can add ctor parameters if needed)
PLAYER = MyPlayer("StudentsPlayer", AUTHORS)
```
