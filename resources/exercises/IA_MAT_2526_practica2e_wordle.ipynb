### Diferencia

Este ejercicio es una ampliación de la Práctica 2, con las siguientes diferencias:
El juego a resolver es **Wordle - with repetition**, no **Mastermind - with repetition**.
Esto significa que el feedback se proporciona de forma distinta.

#### Información a tener en cuenta

- `Wordle` utiliza la clase `WordleFeedback` para representar el feedback de cada guess, mientras `Mastermind` usa `MastermindFeedback`.
    - Feedback en `Wordle` se da con una lista de `WordleFeedback`, que es una enumeración con valores:
        - `WordleFeedback.Wrong = 0`
        - `WordleFeedback.Misplaced = 1`
        - `WordleFeedback.Correct = 2`
    - Feedback en `Mastermind` se da con un objeto `MastermindFeedback` con 2 atributos:
        - `MastermindFeedback.correct: int`
        - `MastermindFeedback.misplaced: int`
- `Wordle` da un feedback específico para cada uno de los valores del código en `guess`, mientras que `Mastermind` da el conteo de estos valores.
